```python

def search_pizza(restaurant_id, pizza_name):
    restaurant = restaurant_data.get(restaurant_id)

    if restaurant is None:
        return "Restaurant not found."

    for pizza in restaurant["items"]:
        if pizza["name"].lower() == pizza_name.lower():
            return pizza

    return "Pizza not found."


# Test the search function
result = search_pizza("pizza_house", "Margherita")

print(result)
```

```python

from sentence_transformers import SentenceTransformer

# Load a multilingual embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

print("Embedding model loaded successfully!")
```

```python

documents = []

for restaurant_id, restaurant in restaurant_data.items():

    # Restaurant general information
    general_info = f"""
    Restaurant: {restaurant['name']}
    Opening hours: {restaurant['opening_hours']}
    Delivery zones and prices: {restaurant['delivery_zones']}
    Promotion: {restaurant['promotion']}
    """

    documents.append({
        "text": general_info,
        "restaurant_id": restaurant_id,
        "type": "general"
    })

    # Pizza information
    for pizza in restaurant["items"]:

        pizza_info = f"""
        Restaurant: {restaurant['name']}
        Pizza: {pizza['name']}
        Prices by size: {pizza['sizes']}
        Ingredients: {pizza['ingredients']}
        Allergens: {pizza['allergens']}
        """

        documents.append({
            "text": pizza_info,
            "restaurant_id": restaurant_id,
            "type": "menu"
        })

print("Documents created:", len(documents))
print("\nExample document:")
print(documents[0]["text"])
```

```python

texts = [doc["text"] for doc in documents]

document_embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embeddings created successfully!")
print("Number of vectors:", len(document_embeddings))
print("Vector dimensions:", document_embeddings.shape[1])
```

```python

import numpy as np

def retrieve_documents(question, restaurant_id=None, top_k=3):

    # Convert the question into an embedding
    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True,
        normalize_embeddings=True
    )[0]

    # Calculate similarity scores
    scores = np.dot(document_embeddings, question_embedding)

    # Filter by restaurant if specified
    candidates = []

    for index, document in enumerate(documents):
        if restaurant_id is None or document["restaurant_id"] == restaurant_id:
            candidates.append((index, scores[index]))

    # Sort by similarity
    candidates.sort(key=lambda x: x[1], reverse=True)

    # Return the most relevant documents
    results = []

    for index, score in candidates[:top_k]:
        results.append({
            "text": documents[index]["text"],
            "score": float(score),
            "restaurant_id": documents[index]["restaurant_id"]
        })

    return results


print("Retrieval function created successfully!")
```

```python

question = "How much is a Large Margherita pizza?"

results = retrieve_documents(
    question,
    restaurant_id="pizza_house",
    top_k=3
)

for i, result in enumerate(results, start=1):
    print(f"\nResult {i}")
    print("Similarity:", round(result["score"], 3))
    print(result["text"])
    print("-" * 50)
```

```python

def build_rag_prompt(question, results):

    context = "\n\n".join(
        result["text"] for result in results
    )

    prompt = f"""
You are a helpful multilingual pizza ordering assistant.

Answer the customer's question using ONLY the information
provided in the context.

Rules:
1. Never invent prices, ingredients, or restaurant information.
2. If the answer is not in the context, say you do not know.
3. Answer in the same language as the customer.
4. Be polite and concise.

CONTEXT:
{context}

CUSTOMER QUESTION:
{question}

ANSWER:
"""

    return prompt


# Test the prompt
question = "Does Tuna Special contain fish?"

results = retrieve_documents(
    question,
    restaurant_id="italian_corner",
    top_k=2
)

prompt = build_rag_prompt(question, results)

print(prompt)
```
