```python

import json

# Load the restaurant database
with open("data/restaurants.json", "r", encoding="utf-8") as file:
    restaurant_data = json.load(file)

print("Database loaded successfully!\n")

# Check the restaurants
for restaurant_id, restaurant in restaurant_data.items():
    print("Restaurant:", restaurant["name"])
    print("Number of pizzas:", len(restaurant["items"]))
    print("Opening hours:", restaurant["opening_hours"])
    print("-" * 40)
```

```python

def search_pizza(restaurant_id, pizza_name):
    restaurant = restaurant_data.get(restaurant_id)

    if restaurant is None:
        return "Restaurant not found."

    for pizza in restaurant["items"]:
        if pizza["name"].lower() == pizza_name.lower():
            return pizza

    return "Pizza not found."


# Test the search function
result = search_pizza("pizza_house", "Margherita")

print(result)
```
