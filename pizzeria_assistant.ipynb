```python

def search_pizza(restaurant_id, pizza_name):
    restaurant = restaurant_data.get(restaurant_id)

    if restaurant is None:
        return "Restaurant not found."

    for pizza in restaurant["items"]:
        if pizza["name"].lower() == pizza_name.lower():
            return pizza

    return "Pizza not found."


# Test the search function
result = search_pizza("pizza_house", "Margherita")

print(result)
```

```python

from sentence_transformers import SentenceTransformer

# Load a multilingual embedding model
embedding_model = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

print("Embedding model loaded successfully!")
```

```python

documents = []

for restaurant_id, restaurant in restaurant_data.items():

    # Restaurant general information
    general_info = f"""
    Restaurant: {restaurant['name']}
    Opening hours: {restaurant['opening_hours']}
    Delivery zones and prices: {restaurant['delivery_zones']}
    Promotion: {restaurant['promotion']}
    """

    documents.append({
        "text": general_info,
        "restaurant_id": restaurant_id,
        "type": "general"
    })

    # Pizza information
    for pizza in restaurant["items"]:

        pizza_info = f"""
        Restaurant: {restaurant['name']}
        Pizza: {pizza['name']}
        Prices by size: {pizza['sizes']}
        Ingredients: {pizza['ingredients']}
        Allergens: {pizza['allergens']}
        """

        documents.append({
            "text": pizza_info,
            "restaurant_id": restaurant_id,
            "type": "menu"
        })

print("Documents created:", len(documents))
print("\nExample document:")
print(documents[0]["text"])
```

```python

texts = [doc["text"] for doc in documents]

document_embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embeddings created successfully!")
print("Number of vectors:", len(document_embeddings))
print("Vector dimensions:", document_embeddings.shape[1])
```

```python

import numpy as np

import time
from google.genai import errors


def ask_restaurant_assistant(
    question,
    restaurant_id,
    model_name="gemini-3.1-flash-lite",
    top_k=3,
    max_retries=4,
):
    normalized_question = normalize_question(question)

    results = retrieve_documents(
        normalized_question,
        restaurant_id=restaurant_id,
        top_k=top_k,
    )

    prompt = build_rag_prompt(question, results)

    for attempt in range(max_retries):
        try:
            response = client.models.generate_content(
                model=model_name,
                contents=prompt,
            )
            return response.text

        except errors.APIError as e:
            if e.code in (429, 503) and attempt < max_retries - 1:
                wait = 6 * (attempt + 1)
                print(f"API status {e.code}. Retrying in {wait}s...")
                time.sleep(wait)
            else:
                raise
```

```python

def build_rag_prompt(question, results):

    context = "\n\n".join(
        result["text"] for result in results
    )

    prompt = f"""
You are a helpful multilingual pizza ordering assistant.

Answer the customer's question using ONLY the information
provided in the context.

IMPORTANT LANGUAGE RULES:
1. If the customer writes in English, answer in English.
2. If the customer writes in French, answer in French.
3. If the customer writes in Algerian Darija using Arabic letters,
   understand the question and answer in Algerian Darija using Arabic letters.
4. If the customer writes in Arabizi (Algerian Darija using Latin
   letters and numbers), understand the question and answer in Arabizi.
5. Do not translate the answer into another language unless requested.

INFORMATION RULES:
1. Never invent prices, ingredients, allergens, or restaurant information.
2. Use ONLY the information provided in the context.
3. If the requested information is missing, clearly say that you do not know.
4. For allergy questions, state ONLY the allergens listed in the context and explicitly note that cross-contamination cannot be ruled out and safety cannot be guaranteed.
5. Be polite, clear, and concise.

CONTEXT:
{context}

CUSTOMER QUESTION:
{question}

ANSWER:
"""

    return prompt
```

```python
def normalize_question(question):

    prompt = f"""
Translate the following customer question into English.

Rules:
- Understand Algerian Darija.
- Understand Arabizi.
- Understand French.
- Preserve pizza names and sizes.
- Do not answer the question.
- Only return the translated question.

Customer question:
{question}

English translation:
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite",
        contents=prompt
    )

    return response.text.strip()
```

```python
smoke_cases = [
    # 1. اسم مطعم بأحرف كبيرة ومسافات (كان يفشل سابقاً)
    ("italian_corner", "Does Classic Margherita contain milk?"),
    ("Italian Corner", "What are the delivery zones and prices?"),

    # 2. لغات متعددة: إنجليزية، فرنسية، دارجة عربية، عربيزي
    ("pizza_house", "How much is a Large Pepperoni?"),
    ("pizza_house", "Quel est le prix d'une grande pizza Margherita ?"),
    ("pizza_house", "شحال تسوى بيتزا مارغريتا كبيرة؟"),
    ("pizza_house", "Pizza Margherita kbira bch7al?"),

    # 3. عنصر غير موجود في القائمة (يجب أن يرفض دون اختراع)
    ("fast_pizza", "Do you have sushi or pasta on the menu?"),

    # 4. سؤال حساسية (يجب ألا يضمن الأمان المطلق)
    ("pizza_house", "Is the Large Margherita safe for someone with a peanut allergy?")
]

for rest_id, q in smoke_cases:
    print(f"Restaurant: {rest_id}")
    print(f"Question : {q}")
    try:
        ans = ask_restaurant_assistant(q, rest_id)
        print(f"Answer   : {ans.strip()}")
    except Exception as e:
        print(f"Error    : {type(e).__name__} -> {e}")
    print("-" * 65)
```

```python
import json
import time
from google.genai import errors

with open("tests/menu_tests.json", "r", encoding="utf-8") as f:
    saved_menu_tests = json.load(f)

for item in saved_menu_tests:
    try:
        pred = ask_restaurant_assistant(item["question"], item["restaurant"]).strip()
        print(f"[{item['id']}] Q: {item['question']}")
        print(f"      Expected : {item['expected_answer']}")
        print(f"      Predicted: {pred}")
    except errors.APIError as e:
        print(f"[{item['id']}] API Error ({e.code}): {e}")
    except Exception as e:
        print(f"[{item['id']}] Error: {type(e).__name__} -> {e}")

    print("-" * 60)
    time.sleep(5)
```
